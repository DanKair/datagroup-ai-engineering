Главная задача: из «грязного» текста собрать сводку по городам. То, что мы разбирали на занятии, только теперь целиком самостоятельно.

Строка хорошая, если в ней ровно три поля и баллы состоят только из цифр. Остальные строки плохие. Города приводите к виду «С большой буквы», без пробелов по краям: « астана » и «АСТАНА» это один город «Астана».

* `bad_rows`: сколько строк данных оказалось плохими (заголовок не считается);
* `city_avg`: **словарь** «город: средний балл», средний округлите до одного знака;
* `best_city`: город с самым высоким средним баллом.

**Направление.** Разбейте задачу на три шага и проверяйте каждый через `print`: разобрать строки и отсеять плохие, сгруппировать баллы по городам в словарь списков, посчитать средние.

**Проверьте себя.** Посмотрите глазами на плохие строки и объясните, чем плоха каждая. Их должно быть ровно столько, сколько насчитала программа.

```python
raw = """name;city;score
Айгерим;Караганда;87
ерлан;  астана ;92
Дана;Алматы;
Тимур;шымкент;78
Бақыт;Астана;74
әлия;АСТАНА;95
Марат;алматы;сто
Жанна;Алматы;81
Нурлан;Караганда;66
Сауле;Алматы;90;лишнее
Асель;Шымкент;84
Руслан;Астана;-5
"""

# Note: For city cleaning use capitalize() method
# 1. Seperate row string like csv files using semi-colon delimiter
# 2. To find city_avg and best_city use city_score dict

raw = raw.split("\n")
city_score = {}

bad_rows = 0

# Starting from index 1 cause first zero index stores: name, city, score headers
# Ending at -1 because last element that stores empty string list item shouldn't be included
for row in raw[1:-1]:
  row = row.split(";")
  # print(row)

  # Catches bad data: not int value / empty string in `score` and if row numbers exceeds
  if len(row) != 3 or not row[2].isdigit() or row[2] == '':
    # print(row)  # Useful to catch bad strings
    bad_rows += 1
    continue

  city = row[1].strip().capitalize()
  score = int(row[2])

  # Using 'dict as counter' technique from our lecture 3 material
  # Note: instead of int `count` we use list, to which we append new item and compute avg using sum(list) / len(list)
  if city not in city_score:
    city_score[city] = [score]
  else:
    city_score[city].append(score)

city_avg = {}
for city, scores in city_score.items():
  city_avg[city] = round(sum(scores) / len(scores), 1)

# Находим город с максимальным средним баллом напрямую из city_avg
best_city = max(city_avg, key=city_avg.get)

print("плохих строк:", bad_rows)
print(city_avg)
print("лучший город:", best_city)
```

* `avg_score`: средний балл **всей** группы, округлённый до одного знака.

**Направление.** Отсортировать имена по баллам можно через список пар `(балл, имя)`: пары сравниваются по первому элементу.

**Проверьте себя.** Поменяйте порог на 90: останутся Жанна и Дана, именно в таком порядке.

```python
students = [
    {"name": "Айгерим", "score": 87},
    {"name": "Ерлан", "score": 64},
    {"name": "Дана", "score": 92},
    {"name": "Тимур", "score": 73},
    {"name": "Әлия", "score": 58},
    {"name": "Марат", "score": 79},
    {"name": "Жанна", "score": 95},
    {"name": "Бақыт", "score": 69},
]

# Фильтруем студентов со счетом >= 70 в виде списка кортежей (балл, имя)
passed_pairs = []
for student in students:
  if student["score"] >= 70:
    passed_pairs.append((student["score"], student["name"]))

# Сортируем кортежи по убыванию (сначала сравниваются баллы)
passed_pairs.sort(reverse=True)

# Извлекаем только имена из отсортированного списка
passed = [name for score, name in passed_pairs]

avg_score = round(sum(student["score"] for student in students) / len(students), 1)

print(passed)
print(avg_score)
```
